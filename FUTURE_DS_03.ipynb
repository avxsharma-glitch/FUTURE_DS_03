```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load YOUR dataset
df = pd.read_csv('Marketing_Funnel_Task3.csv')
print("Dataset shape:", df.shape)
print("\nFunnel stages:")
print(df['Final Stage'].value_counts())
print("\nFirst 5 rows:")
print(df.head())
```

```python
# Funnel summary
funnel = df['Final Stage'].value_counts().sort_index()
conversion_rate = funnel / len(df) * 100

print("Funnel Breakdown:")
print(funnel)
print("\nConversion Rates (%):")
print(conversion_rate.round(1))
```

```python
plt.figure(figsize=(10, 6))
plt.barh(funnel.index, funnel.values, color='skyblue')
plt.title('Marketing Funnel - Task 3 (Aryan Vishwakarma)', fontsize=16, fontweight='bold')
plt.xlabel('Number of Leads')
for i, v in enumerate(funnel.values):
    plt.text(v + 10, i, str(v), va='center')
plt.tight_layout()
plt.savefig('funnel_chart.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
print("\n🎯 BUSINESS INSIGHTS:")
print("1. Lead → MQL drop-off:", round(100 - conversion_rate['MQL'], 1), "%")
print("2. Overall conversion:", round(conversion_rate['Closed Won'], 1), "%")
print("3. Revenue potential:", df['Revenue'].sum())
```
